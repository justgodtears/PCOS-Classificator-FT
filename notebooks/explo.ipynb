```python
import collections
from collections import Counter
from pathlib import Path
import os

# Two classess in dataset
# infected & noninfected

dataset_path = Path(r"C:\Users\justg\Desktop\Development\PCOS FT\dataset\raw")
infected_path = Path(dataset_path, "infected")
noninfected_path = Path(dataset_path, "noninfected")

print(f"Total non infected: {sum(1 for x in noninfected_path.iterdir())}")
print(f"Total infected: {sum(1 for x in infected_path.iterdir())}")
```

```python
from PIL import Image


all_images = [f for f in dataset_path.rglob("*") if f.is_file()]
all_images_size = [Image.open(f).size for f in all_images]
min_images = min(all_images_size)
max_images = max(all_images_size)
print(f"Min images: {min_images}")
print(f"Max images: {max_images}")

print(f"Most common sizes:{Counter(all_images_size).most_common(3)}")

counter = {}
for f in all_images:
    mode = Image.open(f).mode
    if mode in counter:
        counter[mode] += 1
    else:
        counter[mode] = 1

print(counter)
```

```python
import csv
import random

random.seed(42)
imgs_list = random.sample(all_images, 200)

with open(r"C:\Users\justg\Desktop\Development\PCOS FT\dataset\sample_imgs\data.csv", "w", newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    for img in imgs_list:
        writer.writerow([img])

```
